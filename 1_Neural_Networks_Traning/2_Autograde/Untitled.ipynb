```python
import torch
```

```python
material_cost=torch.tensor(10,requires_grad=True,dtype=torch.float64)
labour_cost=torch.tensor(5,requires_grad=True,dtype=torch.float64)
```

```python
total_cost=3 * material_cost**2 + 5 *  labour_cost + 100
```

```python
total_cost.backward()
material_cost.grad, labour_cost.grad
```

```python
x=torch.tensor(4,requires_grad=True,dtype=torch.float64)
y=x ** 2 +100
y.backward()
x.grad
```

```python
with torch.no_grad():
    y= 2 * x ** 3 +7
    print(y.requires_grad)
```

```python
x=torch.tensor([[1,2],[4,3]],requires_grad=True,dtype=torch.float64)
y= 2 * x ** 3 +7
result=y.sum()
result.backward()
```
