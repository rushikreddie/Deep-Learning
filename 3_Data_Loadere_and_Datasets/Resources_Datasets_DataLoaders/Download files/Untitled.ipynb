```python
import torch
from torchvision import datasets
from torchvision.transforms import ToTensor
```

```python
traning_data=datasets.FashionMNIST(
    root="data",
    download=True,
    train=True,
    transform=ToTensor()
)
test_data=datasets.FashionMNIST(
    root="data",
    download=True,
    train=False,
    transform=ToTensor()
)
```

```python
labels_map = {
    0: "T-Shirt",
    1: "Trouser",
    2: "Pullover",
    3: "Dress",
    4: "Coat",
    5: "Sandal",
    6: "Shirt",
    7: "Sneaker",
    8: "Bag",
    9: "Ankle Boot",
}

labels_map[label]
```

```python
figure = plt.figure(figsize=(8, 8))
cols, rows = 3, 3
for i in range(1, cols * rows + 1):
    sample_idx = torch.randint(len(traning_data), size=(1,)).item()
    img, label = traning_data[sample_idx]
    figure.add_subplot(rows, cols, i)
    plt.title(labels_map[label])
    plt.axis("off")
    plt.imshow(img.squeeze(), cmap="gray")
plt.show()
```

```python
from torch.utils.data import DataLoader
```

```python
train_loader=DataLoader(traning_data,batch_size=64,shuffle=True)
test_loader=DataLoader(test_data,batch_size=64,shuffle=True)
```

```python
data_iter=iter(train_loader)
image,labels=next(data_iter)
```
