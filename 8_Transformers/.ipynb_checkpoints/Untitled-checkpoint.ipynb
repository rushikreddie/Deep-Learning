```python
from transformers import BertTokenizer, BertModel
```

```python
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
```

```python
tokens=encoded_input = tokenizer("i love apple ", return_tensors='pt')
tokens
```

```python
tokens=encoded_input = tokenizer(["i love apple ","you are what you eat"],padding=True,return_tensors='pt')
tokens
```

```python
tokens=encoded_input = tokenizer(["i love apple ","you are what you eat"],padding="max_length",max_length=10,return_tensors='pt')
tokens
```

```python
tokens=encoded_input = tokenizer(["i love apple ","you are what you eat"],padding="max_length",max_length=5,truncation=True,return_tensors='pt')
tokens
```

```python
tokens=encoded_input = tokenizer("i love apple ", return_tensors='pt')
tokens
```

```python
model = BertModel.from_pretrained("bert-base-uncased")
output=model(**tokens)
```

```python
tokenizer.convert_ids_to_tokens(tokens["input_ids"][0])
```
